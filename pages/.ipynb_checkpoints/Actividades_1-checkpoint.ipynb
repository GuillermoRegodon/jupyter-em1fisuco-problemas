## 1. El Principio de Superposición

El Principio de Superposición constituye el pilar operativo central de la electrostática. Su validez es una consecuencia directa de la linealidad de las ecuaciones fundamentales que rigen el campo eléctrico en el vacío (las ecuaciones de Maxwell en régimen estático). En particular, **la expresión general integral para el potencial escalar en un punto de observación del espacio definido por el vector de posición $\vec{r}$ es lineal**:

$$
\varphi(\vec{r}) = \frac{1}{4\pi\varepsilon_0} \int_{V'} \frac{\rho(\vec{r}')}{|\vec{r} - \vec{r}'|} dV'
$$

donde $\rho(\vec{r}')$ representa la densidad volumétrica de carga en la posición de la fuente $\vec{r}'$, $dV'$ es el elemento diferencial de volumen en el espacio de la fuente, y $|\vec{r} - \vec{r}'|$ es la distancia euclídea entre la fuente y el observador. Observamos que la integral se puede dividir en dos sentidos:

* Se puede dividir el dominio de integración, $V' = V_1' \bigcup V_2'$ (*con su intersección $V_1' \bigcap V_2' = \emptyset$ el conjunto vacio*):

$$
\frac{1}{4\pi\varepsilon_0} \int_{V'} \frac{\rho(\vec{r}')}{|\vec{r} - \vec{r}'|} dV' = \frac{1}{4\pi\varepsilon_0} \int_{V'_1} \frac{\rho(\vec{r}')}{|\vec{r} - \vec{r}'|} dV' + \frac{1}{4\pi\varepsilon_0} \int_{V'_2} \frac{\rho(\vec{r}')}{|\vec{r} - \vec{r}'|} dV'
$$

* Se puede descomponer $\rho(\vec{r}') = \rho_1(\vec{r}') + \rho_2(\vec{r}')$ en el mismo volumen:

$$
\frac{1}{4\pi\varepsilon_0} \int_{V'} \frac{\rho(\vec{r}')}{|\vec{r} - \vec{r}'|} dV' = \frac{1}{4\pi\varepsilon_0} \int_{V'} \frac{\rho_1(\vec{r}')}{|\vec{r} - \vec{r}'|} dV' + \frac{1}{4\pi\varepsilon_0} \int_{V'} \frac{\rho_2(\vec{r}')}{|\vec{r} - \vec{r}'|} dV'
$$


```{note}
Es crucial comprender que el principio de superposición **solo es estrictamente aplicable de forma directa cuando las distribuciones de carga son rígidas, fijas o predeterminadas**. La presencia de Conductores Eléctricos Perfectos (CEP) introduce una interacción no lineal con el entorno: en un CEP, las cargas no están fijadas a una posición geométrica estática, sino que se redistribuyen libremente sobre la superficie del conductor en respuesta al campo eléctrico externo aplicado. En estos escenarios, el campo externo modifica la densidad de carga inducida y esta nueva distribución alterada modifica a su vez el campo total resultante. Por tanto, no se pueden sumar directamente los campos sin resolver previamente el estado de equilibrio de la carga superficial inducida.
```

### Traducción de Integrales Físicas a Integrales Matemáticas Calculables

Utilizar la expresión general integral para el potencial escalar requiere transformar una expresión física abstracta en una integral matemática definida resoluble en el sistema de coordenadas apropiado. El algoritmo metodológico estructurado consta de 4 pasos fundamentales:

* **Parametrización del vector de posición del punto de observación ( $\vec{r}$ )**:  Definir analíticamente las coordenadas del punto del espacio donde se desea calcular el potencial o el campo eléctrico.

* **Parametrización del vector de posición de la fuente ( $\vec{r}'$ )**:  Expresar la posición de un elemento genérico diferencial de carga en función de las variables de integración.

* **Definición del elemento diferencial apropiado ( $dl'$ , $dS'$ , o $dV'$ )**:  Especificar el diferencial geométrico adecuado según la dimensión de la fuente cargada.

* **Establecimiento de los límites de integración**:  Fijar las cotas numéricas o analíticas que barren completamente la geometría de la fuente.


Mediante este protocolo, la integral física abstracta del campo eléctrico:

$$\vec{E}(\vec{r}) = \frac{1}{4\pi\varepsilon_0} \int_{V'} \frac{\rho(\vec{r}')(\vec{r} - \vec{r}')}{|\vec{r} - \vec{r}'|^3} dV'$$

o de la fuerza resultante sobre una distribución:

$$\vec{F} = \int_{V} \rho(\vec{r}) \vec{E}(\vec{r}) dV$$

se transforma en integrales vectoriales o escalares bien definidas evaluables mediante el cálculo infinitesimal ordinario.



### Actividades Prácticas

* Actividad 1.1 (Hilo horizontal desplazado): Considérese un hilo rectilíneo continuo de longitud $L$ cargado uniformemente con densidad lineal constante $\lambda$, paralelo al eje $y$ desplazado en la dirección del eje $x$. ¿Cuál es el potencial electrostático en un punto del eje $z$ dado por $\vec{r} = (0,0,b)$?

```{figure}../images/A1_1.png
:alt: A1_1.png
:width: 250px
```




## 2. Desarrollo Multipolar del Potencial Electrostático y Análisis de Escalas
El desarrollo multipolar es una herramienta util para aproximar el potencial y el campo eléctrico a distancias grandes en comparación con las dimensiones geométricas de la distribución de carga. La idea física subyacente consiste en expandir el campo en una serie infinita de términos organizados por simetrías vectoriales y potencias inversas de la distancia ( $1/r^n$ ). **Desde el punto de vista práctico, el valor esencial del método radica en poder de truncar la serie tras los primeros términos dominantes, despreciando el resto cuando $r$ es suficientemente grande**.


### Formalismo Matemático del Desarrollo Multipolar

El potencial electrostático $\varphi(\vec{r})$ creado por una distribución acotada de carga en puntos lejanos ( $r \gg r'$ ) se expande en serie de Taylor/Legendre como:

$$
\varphi(\vec{r}) = \frac{1}{4\pi\varepsilon_0} \left(\frac{Q_T}{r} + \frac{\vec{r} \cdot \vec{p}}{r^3} + \frac{1}{2} \frac{\vec{r} \cdot \overleftrightarrow{Q} \cdot \vec{r}}{r^5} + \dots \right)
$$

donde:

* $Q_T = \int \rho(\vec{r}') dV'$ es la carga total o momento monopolar (escalar).
* $\vec{p} = \int \vec{r}' \rho(\vec{r}') dV'$ es el momento dipolar eléctrico (vector).
* $\overleftrightarrow{Q}$ es el tensor momento cuadrupolar eléctrico.

Al truncar la serie tras el término cuadrupolar, el orden del error asociado en el campo lejano es $\mathcal{O}(1/r^4)$. **El campo eléctrico se deriva aplicando el gradiente negativo ( $\vec{E} = -\nabla\varphi$ )**. Para los casos de términos monopolar y dipolar, las expresiones son:

* Campo Monopolar:  Derivado inmediatamente de $Q_T/r$ :

$$
\vec{E}_{\text{mon}}(\vec{r}) = \frac{Q_T}{4\pi\varepsilon_0 r^2} \hat{\vec{r}}
$$

* Campo Dipolar:  Obtenido a partir de $\frac{\vec{r} \cdot \vec{p}}{r^3}$ :

$$
\vec{E}_{\text{dip}}(\vec{r}) = \frac{1}{4\pi\varepsilon_0} \left( \frac{3\vec{r}(\vec{p} \cdot \vec{r})}{r^5} - \frac{\vec{p}}{r^3} \right)
$$


### Criterios de Aplicabilidad y Comparación de Escalas

La validez del truncamiento multipolar no depende de la magnitud absoluta de la distancia $r$, **sino de la comparación de escalas relativas**. La distancia del observador debe ser sensiblemente mayor que la dimensión característica $L$ de la fuente ( $r \gg L$ ). Ponemos tres ejemplos que pueden servir para entender la comparación de escalas:

* **Efecto de un dipolo sobre un objeto grande**: Conociendo el campo dipolar $\vec{E}_{\text{dip}}(\vec{r})$ en la región lejana, la fuerza neta sobre un objeto extenso cargado con densidad $\rho(\vec{r})$ sumergido en dicho campo viene dada por la integración volumétrica:

$$
\vec{F} = \int_{V} \rho(\vec{r}) \vec{E}_{\text{dip}}(\vec{r}) dV
$$

* **Efecto de un objeto grande sobre un dipolo**: Asumiendo que el tamaño del dipolo $d$ es mucho menor que la escala $L$ del objeto grande ( $d \ll L$ ), el campo eléctrico $\vec{E}$ generado por el objeto se percibe cuasi-uniforme sobre el dipolo. La fuerza y el momento de fuerza resultantes son:

$$
\vec{F} = (\vec{p} \cdot \nabla)\vec{E} \\
\vec{M} = \vec{p} \times \vec{E}
$$

* **Interacción Dipolo-Dipolo**: Para dos dipolos arbitrarios $\vec{p}_1$ y $\vec{p}_2$ separados por una distancia $a = |\vec{r}_2- \vec{r}_2|$ mucho mayor que sus tamaños ( $a \gg d_1, d_2$ ), el dipolo 1 genera un campo $\vec{E}_1(\vec{r}_2- \vec{r}_1)$ en la posición del dipolo 2.


### Actividades Prácticas

* Actividad 2.1 (Geometría de hilos contrapuestos): Tenemos un sistema de dos segmentos colineales de longitud $b$ separados una distancia $2a$. Un segmento posee densidad $+\lambda$ y el otro $-\lambda$. ¿Cuáles son los momentos multipolares de orden 0 y 1 (Monopolar y dipolar)?

```{figure}../images/A1_3.png
:alt: A1_3.png
:width: 250px
```

### Obtención de la Carga Real Inducida

La distribución física de carga sobre la superficie del conductor se obtiene aplicando la condición de frontera en la interfaz conductor-vacío:

$$
\varepsilon_0 (\vec{E}_2 - \vec{E}_1) = \sigma \hat{\vec{n}}
$$

Dado que en el interior del conductor el campo es nulo ( $\vec{E}_1 = \vec{0}$ ), la densidad superficial de carga inducida $\sigma(\vec{r}_{\text{sup}})$ viene dada directamente por la componente normal del campo exterior $\vec{E}_2$ :

$$
\sigma(\vec{r}_{\text{sup}}) = \varepsilon_0 E_{2,n}
$$

El procedimiento completo requiere:

* Calcular el campo exterior $\vec{E}_2$ mediante la superposición del campo **de la carga real exterior y el de las cargas imagen**.

* (*Opcional*) Comprobar que $\vec{E}_2$ sea estrictamente normal a la superficie conductora. Para CEPs, debe serlo si está bien resuelto.

* Despejar $\sigma(\vec{r}_{\text{sup}})$, que **casi nunca es uniforme**.


### Sistemas Fundamentales con Cargas Puntuales

Para garantizar exactamente la condición de contorno $\varphi = 0$ en la superficie del CEP:

* **Plano Infinito de CEP a Tierra ( $\varphi=0$ )**.

    * Carga imagen: $q' = -q$.
    * Posición imagen: $d' = d$, ubicada en la posición simétrica reflejada detrás del plano.


## 4. Método de las Imágenes II: Distribuciones Continuas de Carga y Configuraciones Complejas

El tratamiento de distribuciones continuas de carga $\lambda$, $\sigma$ o $\rho$ mediante el método de las imágenes se fundamenta en considerar la distribución como un conjunto infinito de cargas puntuales diferenciales $dq$. La reflexión de cada $dq$ genera un elemento diferencial de carga imagen $dq'$, cuya superposición construye la densidad de carga imagen continua en el dominio reflejado. La diferencia fundamental es que las cargas puntuales no ocupan espacio (*solo un punto*), mientras que los diferenciales de carga distribuida ocupan un espacio $dl$, $dS$ o $dV$, y debemos tener cuidado con comprobar qué espacio ocupa la imagen.


### Distribuciones Continuas frente a Planos Conductores

Para un plano conductor infinito a tierra, la simetría especular del espacio diferencial es directa: cada diferencial $dq$ situado a una distancia $d$ genera un diferencial imagen $dq'$ a la misma distancia detrás del plano. Dado que la reflexión plana preserva los elementos geométricos de área y longitud ( $dV = dV'$, $dS = dS'$, $dl = dl'$ ):

$$
dq' = -dq \implies \rho' dV' = -\rho dV \implies \rho' = -\rho \\
dq' = -dq \implies \sigma' dS' = -\sigma dS \implies \sigma' = -\sigma \\
dq' = -dq \implies \lambda' dl' = -\lambda dl \implies \lambda' = -\lambda
$$

```{note}
Observamos que tenemos dos pares de ecuaciones:

* Las definiciones de $dq = \rho dV = \sigma dS = \lambda dl$ y $dq' = \rho' dV' = \sigma' dS' = \lambda' dl'$.

* Las ecuaciones de las imagenes $dq' = -dq$ y $d = d'$.

Además de la condicion de igualdad entre los tamaños de los diferenciales $dV = dV'$, $dS = dS'$ y $dl = dl'$.
```

### Distribuciones Continuas frente a Esferas Conductoras de Radio $a$

En geometrías esféricas, la transformación de posición $d' = \frac{a^2}{d}$ no es lineal, lo que introduce deformaciones diferenciales en la geometría de la imagen. A partir de la relación diferencial $dq' = -\frac{a}{d} dq$, sustituimos $dq = \lambda dl$ y $dq' = \lambda' dl'$:

$$
\lambda' dl' = -\frac{a}{d} \lambda dl \implies \lambda' = -\frac{a}{d} \lambda \left| \frac{dl}{dl'} \right|
$$

Observamos primero que las cargas $dq = \lambda dl$ y $dq' = \lambda' dl'$ dependen de la longitud del intervalo, por lo que ponemos un valor absoluto para conservar la definición física. La densidad de carga imagen $\lambda'$ depende de manera crítica de la orientación espacial del hilo fuente respecto a la esfera conductora:

* **Hilo en Orientación Transversal** (distancia $b$ ):  Para un hilo circular situado a un radio constante $r = b$ del centro:

    * Posición de la imagen: $b' = \frac{a^2}{b}$ (constante).
    * Diferenciales: $dl = b d\alpha$ y $dl' = b' d\alpha = \frac{a^2}{b} d\alpha$. Observamos que la geometría dicta que el ángulo debe ser el mismo, por que ambas imagenes están en el mismo eje radio-carga que sus cargas originales.
    * Razón de diferenciales: $\left| \frac{dl}{dl'} \right| = \frac{b}{b'} = \frac{b^2}{a^2}$.
    * Densidad transformada:

$$
\lambda' = -\frac{a}{b} \lambda \left( \frac{b^2}{a^2} \right) = -\frac{b}{a} \lambda
$$

```{figure}../images/A1_7.png
:alt: A1_7.png
:width: 250px
```


* **Hilo en Orientación Radial**:  Para un hilo recto extendido radialmente desde $r = b$ hasta $r = c$ :
    * Posición de la imagen: $l' = \frac{a^2}{l}$, donde $l$ es la coordenada radial de la fuente.
    * Razón de diferenciales: Derivando $l' = \frac{a^2}{l}$ respecto a $l$, obtenemos $\frac{dl'}{dl} = -\frac{a^2}{l^2} \implies \left| \frac{dl}{dl'} \right| = \frac{l^2}{a^2}$.
    * Sustituyendo en la ecuación de mapeo: $\lambda' = -\frac{a}{l} \lambda \left( \frac{l^2}{a^2} \right) = -\frac{l}{a} \lambda$
    * Expresada en función de la coordenada real del hilo $l$ : $\lambda'(l) = -\frac{l}{a} \lambda$.
    * Expresada en función de la propia coordenada de la imagen $l'$ (mediante $l = \frac{a^2}{l'}$ ):

$$
\lambda'(l') = -\frac{a^2/l'}{a} \lambda = -\frac{a}{l'} \lambda
$$

```{figure}../images/A1_8.png
:alt: A1_8.png
:width: 300px
```

La densidad de carga imagen $\lambda'$ de un hilo radial no es uniforme; varía inversamente con la posición de la imagen $l'$, concentrando una densidad de carga progresivamente mayor en el extremo más cercano a la superficie del conductor.

Obviamente, una distribución plana tiene dos direcciones y una volumétrica tres direcciones que se convertirían usando este mismo principio.

```{note}
Observamos que tenemos tres pares de ecuaciones:

* Las definiciones de $dq = \rho dV = \sigma dS = \lambda dl$ y $dq' = \rho' dV' = \sigma' dS' = \lambda' dl'$.

* Las ecuaciones de las imagenes $dq' = -\frac{a}{d} dq$ y $d = \frac{a^2}{d'}$.

* La relación entre los diferenciales $dV = dV'$, $dS = dS'$ y $dl = dl'$, que es distinta si tiene dirección radial o tranversal.

```

### Efecto de las Condiciones Globales en la Esfera Conductora

El plano de CEP lo conectaremos siempre a tierra, porque es infinito y ponemos la referencia de potencial en el infinito. **La esfera de CEP tiene 3 posibilidades**:

* **Esfera a Tierra** ( $\varphi = 0$ ):
    * Carga imagen $q' = -q\frac{a}{d}$ en $d' = \frac{a^2}{d}$.
    * Carga total sobre la esfera: $Q_{\text{CEP}} = q'$.

* **Esfera a Potencial Fijo** ( $\varphi = V_0$ ):
    * Carga $q'$ en $d'$, más una segunda carga ficticia $Q_C = 4\pi\varepsilon_0 a V_0$ en el centro de la esfera ( $r=0$ ).
    * Carga total sobre la esfera: $Q_{\text{CEP}} = Q_C + q'$.

* **Esfera Aislada con Carga Neta** $Q_0$:
    * Carga $q'$ en $d'$, más una carga central ficticia en $r=0$ de valor $Q_C = Q_0 - q' = Q_0 + q\frac{a}{d}$.
    * Potencial resultante en la esfera: $\varphi_{\text{CEP}} = \frac{Q_C}{4\pi\varepsilon_0 a}$.

Para cualquier distribución continua frente a una esfera de radio $a$, la siguiente tabla sintetiza las cargas ficticias necesarias y las magnitudes globales en los tres estados físicos fundamentales:

| Estado Físico del Conductor | Carga Central Ficticia ( $Q_C$ ) | Potencial en la Superficie ( $\varphi_{\text{CEP}}$ ) |Carga Total Integrada ( $Q_{\text{CEP}}$ ) |
| --- | --- | --- | --- |
| Conectada a Tierra | $Q_C = 0$ | $\varphi_{\text{CEP}} = 0$ | $Q_{\text{CEP}} = \int dq'$ |
| A Potencial Fijo $V$ | $Q_C = 4\pi\varepsilon_0 a V$ | $\varphi_{\text{CEP}} = V$ | $Q_{\text{CEP}} = Q_C + \int dq'$ |
| Aislada con Carga Neta $Q_0$ | $Q_C = Q_0 - \int dq'$ | $\varphi_{\text{CEP}} = \frac{Q_C}{4\pi\varepsilon_0 a}$ | $Q_{\text{CEP}} = Q_0$ |
