## S5. Problemas de valores de contorno y separación de variables en coordenadas esféricas

En un espacio sin densidad volumétrica de carga libre $\rho(\vec{r})$, la ecuación de Poisson se transforma directamente en la ecuación de Laplace. Esta transición matemática preserva el rigor teórico y, al mismo tiempo, **permite obtener soluciones exactas mediante separación de variables y expansiones en serie en funciones de base ortogonales**.

### Transición de la ecuación de Poisson a la ecuación de Laplace

En medios continuos tridimensionales, la distribución espacial del potencial electrostático escalar $\varphi(\vec{r})$ está gobernada por la ecuación diferencial de Poisson:

$$
\nabla^2 \varphi(\vec{r}) = -\frac{\rho(\vec{r})}{\varepsilon_0}
$$

En dominios espaciales donde no hay carga libre, es decir, $\rho(\vec{r}) = 0$, la ecuación de Poisson se reduce a la ecuación de Laplace:

$$
\nabla^2 \varphi(\vec{r}) = 0
$$

Las soluciones de la ecuación de Laplace se conocen como funciones armónicas y poseen propiedades matemáticas notables. Las funciones armónicas son infinitamente diferenciables ($C^\infty$ suaves) dentro de su dominio y cumplen el teorema del valor medio, así como el principio del máximo, que excluye máximos o mínimos locales de potencial en el interior. Son lo que podemos llamar *"Las funciones más suaves posibles"*.

```{note}
Observar que la ecuación de Laplace implica $\rho(\vec{r})$ en todo el espacio. ¿Dónde está la carga que genera el potencial $\varphi$ y el campo eléctrico $\vec{E}$? **En la superficie donde se define la condición de frontera, $\sigma(\vec{r})$**.
```


### Intuición física y analogía con una membrana elástica bidimensional

Para cultivar una comprensión intuitiva de las formulaciones bidimensionales de Laplace, $\nabla^2 h(x,y)=0$, puede examinarse una membrana elástica bidimensional bajo tensión lateral uniforme $\tau$, estirada sobre un contorno rígido de frontera. En el límite de pequeños gradientes, la altura vertical $h(x,y)$ satisface la ecuación de superficie de equilibrio en la que la curvatura media se anula:

$$
H \approx \frac{1}{2}\left(\frac{\partial^2 h}{\partial x^2}+\frac{\partial^2 h}{\partial y^2}\right)=0
\implies
\nabla^2 h(x,y)=0
$$

Aquí, la altura espacial $h(x,y)$ de la membrana constituye un análogo exacto del potencial electrostático $\varphi(x,y)$. La exigencia física de minimizar el área superficial bajo tensión impide la aparición de picos o huecos localizados, lo que obliga al desplazamiento interior a ajustarse suavemente para coincidir con el marco de frontera.

### Metodología de las expansiones en polinomios de Legendre

En dominios tridimensionales con simetría azimutal (*Es decir, que no depende de la variable angular azimutal $\phi$*, o $\frac{\partial{\varphi(r,\theta,\phi)}}{\partial{\phi}}=0$), la ecuación de Laplace expresada en coordenadas esféricas $(r,\theta)$ admite una solución general mediante separación de variables expandida en potencias radiales y polinomios de Legendre $P_l(\cos\theta)$:

$$
\varphi(r,\theta)=\sum_{l=0}^{\infty}\left(A_l r^l+\frac{B_l}{r^{l+1}}\right)P_l(\cos\theta)
$$

donde $A_l$ y $B_l$ son coeficientes de expansión reales **determinados gracias a las condiciones de contorno**. Esta expresión es válida tanto dentro como fuera de la esfera **pero los coeficientes son distintos en los dos volúmenes**. Nos convendrá añadir un subíndice o cambiar la letra para distinguirlos. Sin embargo, ambos tienen la misma condición de frontera (*Aún no la he definido, esperemos un poco antes de concretarla*). Si ambas expresiones son válidas en todo su espacio, y ambos espacios se tocan para $r=R$, los coeficientes dentro y fuera tienen el mismo valor en $r=R$. A ese valor común lo llamamos $C_l$

$$
\varphi(r=R,\theta)&=\sum_{l=0}^{\infty}C_l P_l(\cos\theta) \\
C_l&=A_l R^l+\frac{B_l}{R^{l+1}}
$$

Para calcular $C_l$, para un caso general usariamos la relación de ortogonalidad de los Polinomios de Legendre (*igual que hacemos para obtener los coeficientes de un vector en una base determinada, o los coeficientes de Fourier en un desarrollo en serie de Fourier*). Sin embargo, para poder resolverlo a mano, en nuestros ejercicios vamos a poner siempre condiciones de frontera que se puedan manejar algebráicamente en función de $P_l(\cos\theta)$.

* $P_0(\cos\theta)=1$
* $P_1(\cos\theta)=\cos\theta$
* $P_2(\cos\theta)=\frac{3}{2}\cos^2\theta-\frac{1}{2}$
* etc...

**Vamos a poner la condición de contorno**. Nos encontramos con dos posibilidades:

* La CC viene dada **en función del potencial $\varphi(r=R, \theta)$ en la frontera** (*implica una $\sigma(r=R,\theta)\neq 0$*) (Problema B2.S5.1). Ponemos la condición de frontera en función de $\cos\theta$ y sus potencias y comparamos manualmente los términos para obtener los coeficientes $C_l$ a partir de los datos de la CC. A partir de $C_l$, podemos deducir $A_l$ y $B_l$, escribir el potencial general $\varphi(r, \theta)$, y podemos deducir el campo eléctrico $\vec{E}$ y la densidad superficial de carga en la frontera $\sigma(r=R,\theta)$.

* La CC viene dada **en función de la densidad superficial de carga $\sigma(r=R,\theta)$ en la frontera** (Problema B2.S5.2). La metodología no es invertible, por lo que no tenemos más remedio que utilizar coeficientes $C_l$, $A_l$ y $B_l$ genéricos, calcular el potencial generico $\varphi(r, \theta)$, deducir el campo eléctrico genérico $\vec{E}$ y la densidad superficial de carga genérica en la frontera $\sigma(r=R,\theta)$. Entonces, si podremos comparar los coeficientes genéricos en la densidad superficial de carga genérica en la frontera con la CC $\sigma(r=R,\theta)$ en la frontera.

```{note}
¿Cómo se puede obtener el campo eléctrico $\vec{E}$ a partir del potencial $\varphi(r, \theta)$? ¿Y cómo se puede obtener a densidad superficial de carga en la frontera $\sigma(r=R,\theta)$ a partir del campo eléctrico $\vec{E}$?
```

Para poder definir lo coeficientes $A_l$ y $B_l$, necesitaremos aplicar ciertos criterios de regularidad.

1. **Región interior ($r<R$):** los potenciales físicos deben permanecer finitos en el origen espacial ($r\to0$). Dado que los términos proporcionales a $r^{-(l+1)}$ divergen cuando $r\to0$, la regularidad estructural impone $B_l=0$ para todo $l\ge0$, lo que produce:

   $$
   \varphi_{\text{int}}(r,\theta)=\sum_{l=0}^{\infty}A_l r^l P_l(\cos\theta)
   $$

2. **Región exterior ($r>R$):** para una frontera esférica aislada, el potencial debe decaer suavemente a cero en el infinito ($\lim_{r\to\infty}\varphi=0$). Dado que los términos proporcionales a $r^l$ crecen sin límite para $l>0$ y representan desplazamientos constantes no nulos para $l=0$, la regularidad impone $A_l=0$ para todo $l\ge0$, lo que produce:

   $$
   \varphi_{\text{ext}}(r,\theta)=\sum_{l=0}^{\infty}\frac{B_l}{r^{l+1}}P_l(\cos\theta)
   $$

A partir de esta expresión, podemos resolverlo todo:

$$
\vec{E}(\vec{r})&=-\nabla\varphi(\vec{r}) \\
\sigma(r=R,\theta)&=\varepsilon_0\left(E_{r,\text{ext}}(r=R,\theta)-E_{r,\text{int}}(r=R,\theta)\right)
$$

```{note}
**Aclaración:** Si la esfera está inmersa en un campo externo uniforme $\vec{E}_0=E_0\hat{u}_z$, el potencial en el infinito toma la forma asintótica $-E_0 r\cos\theta$, lo que impone $A_1=-E_0$ y $A_l=0$ para $l\neq1$.
```


### Ejemplo en detalle

Considérese una cáscara esférica hueca de radio $R$ sometida a un potencial superficial prescrito $\varphi(R,\theta)=V_0\cos^2\theta$.

**Paso 1: descomposición en Legendre.** Descomponer $\cos^2\theta$ usando $P_2(\cos\theta)=\frac{3}{2}\cos^2\theta-\frac{1}{2}$ y $P_0(\cos\theta)=1$:

$$
\cos^2\theta=\frac{2}{3}P_2(\cos\theta)+\frac{1}{3}P_0(\cos\theta)
$$

Por tanto, el potencial de contorno se convierte en:

$$
\varphi(R,\theta)=\frac{1}{3}V_0P_0(\cos\theta)+\frac{2}{3}V_0P_2(\cos\theta)
$$

Comparando esto con $\sum C_lP_l(\cos\theta)$ se obtiene $C_0=\frac{1}{3}V_0$, $C_2=\frac{2}{3}V_0$ y $C_l=0$ para $l\neq0,2$.

**Paso 2: soluciones de potencial.**

* **Interior ($r\le R$):** $A_0=C_0=\frac{1}{3}V_0$ y $A_2R^2=C_2\implies A_2=\frac{2V_0}{3R^2}$.

  $$
  \varphi_{\text{int}}(r,\theta)
  =\frac{1}{3}V_0+\frac{2V_0}{3R^2}r^2P_2(\cos\theta)
  =\frac{1}{3}V_0+\frac{V_0r^2}{3R^2}(3\cos^2\theta-1)
  $$

* **Exterior ($r\ge R$):** $\frac{B_0}{R}=C_0\implies B_0=\frac{1}{3}V_0R$ y $\frac{B_2}{R^3}=C_2\implies B_2=\frac{2}{3}V_0R^3$.

  $$
  \varphi_{\text{ext}}(r,\theta)
  =\frac{V_0R}{3r}+\frac{2V_0R^3}{3r^3}P_2(\cos\theta)
  =\frac{V_0R}{3r}+\frac{V_0R^3}{3r^3}(3\cos^2\theta-1)
  $$

**Paso 3: campos eléctricos vectoriales.** Tomando $-\nabla\varphi$ en ambos dominios:

* **Campo interior $\vec{E}_{\text{int}}(r,\theta)$:**

  $$
  E_{r,\text{int}}=-\frac{\partial\varphi_{\text{int}}}{\partial r}
  =-\frac{2V_0r}{3R^2}(3\cos^2\theta-1)
  $$

  $$
  E_{\theta,\text{int}}=-\frac{1}{r}\frac{\partial\varphi_{\text{int}}}{\partial\theta}
  =\frac{2V_0r}{3R^2}\sin\theta\cos\theta
  $$

* **Campo exterior $\vec{E}_{\text{ext}}(r,\theta)$:**

  $$
  E_{r,\text{ext}}=-\frac{\partial\varphi_{\text{ext}}}{\partial r}
  =\frac{V_0R}{3r^2}+\frac{V_0R^3}{r^4}(3\cos^2\theta-1)
  $$

  $$
  E_{\theta,\text{ext}}=-\frac{1}{r}\frac{\partial\varphi_{\text{ext}}}{\partial\theta}
  =\frac{2V_0R^3}{3r^4}\sin\theta\cos\theta
  $$

**Paso 4: densidad superficial de carga $\sigma(\theta)$.** Evaluando los campos radiales en $r=R$:

$$
E_{r,\text{int}}(R,\theta)=-\frac{2V_0}{3R}(3\cos^2\theta-1)
$$

$$
E_{r,\text{ext}}(R,\theta)=\frac{V_0}{3R}+\frac{V_0}{R}(3\cos^2\theta-1)
$$

Restando el campo interior del exterior:

$$
\begin{aligned}
\sigma(\theta)
&=\varepsilon_0\left[
\frac{V_0}{3R}
+\frac{V_0}{R}(3\cos^2\theta-1)
+\frac{2V_0}{3R}(3\cos^2\theta-1)
\right] \\
&=\frac{\varepsilon_0V_0}{R}\left(5\cos^2\theta-\frac{4}{3}\right) \\
&=\frac{\varepsilon_0V_0}{R}\left(\frac{10}{3}P_2(\cos\theta)+\frac{1}{3}P_0(\cos\theta)\right)
\end{aligned}
$$
































## Medios dieléctricos.

Los medios materiales en general se pueden polarizar, lo que implica siempre unas densidades volumétrica y superficial de carga de polarización $\rho_P$ y $\sigma_P$.

$$
\rho_P &= -\nabla \cdot \vec{P} \\
\sigma_P &= \vec{P} \cdot \vec{n}
$$

Consideramos dos posibilidades en relación a cómo se genera la polarización $\vec{P}$ (*o momento dipolar por unidad de volumen*):

* La polarización $\vec{P}$ es proporcional al campo eléctrico $\vec{E}$ (*Seminario 6*).

$$
\vec{P} = \chi_e \vec{E}
$$

* La polarización $\vec{P}$ es constante para cualquier campo eléctrico $\vec{E}$ externo (*Seminario 7*).

$$
\vec{P} = cte \quad \forall \vec{E}
$$


## S6. Medios dieléctricos: Energía electrostática.

Los materiales lineales, $\vec{P} = \chi_e \vec{E}$, se pueden resolver simplificando la relación lineal entre la polarización $\vec{P}$ y el campo eléctrico $\vec{E}$ mediante el concepto de permitividad del material. Por lo tanto, los problemas los resolveremos utilizando la ley de Gauss para materiales, en función del vector desplazamiento:

$$
\oint_S \vec{D}\cdot\vec{dS} = \int_V \rho_l dV = Q_l
$$

Junto con las relaciones:

$$
\vec{D} &= \varepsilon_m\vec{E} \\
\vec{E} &= -\nabla\varphi
$$

Para que esta metodología sea manejable, el problema debe tener simetría (*El material debe tener simetría*). Pero si el problema tiene simetría, significa que lo voy a poder resolver mediante el método de Gauss, eligiendo una superficie conveniente que cumpla las dos condiciones que debe tener (Ver [Repaso](https://guillermoregodon.github.io/jupyter-em1fisuco-problemas/pages/repaso-gauss-ampere/)):

* El campo eléctrico $\vec{E}$ es perperdicular a la superficie.

* El módulo del campo eléctrico $E$ es constante y puede salir de la integral.

Esta metodología se puede usar con cargas determinadas rodeadas de materiales lineales y en condensadores. En el segundo caso, en sistemas de condensador, hay que relacionar la carga total libre acumulada en la superfície conductora con el potencial al que se polariza el condensador y encontrar una relaición proporcional:

$$
C=\frac{Q_l}{V}
$$


### Configuración del sistema de condensador cilíndrico

Considérese un condensador cilíndrico alargado de longitud total $L$, definido por las siguientes especificaciones geométricas y de permitividad:

* **Núcleo conductor interior:** cilindro macizo de radio $r=R/2$ mantenido a potencial eléctrico $V$.
* **Cáscara conductora exterior:** pared cilíndrica delgada de radio $r=R$ conectada a tierra ($V=0$).
* **Subdominio inferior ($0\le z\le h$):** lleno con un líquido dieléctrico lineal de permitividad $\varepsilon$.
* **Subdominio superior ($h<z\le L$):** lleno con un medio de vacío de permitividad $\varepsilon_0$.

```{figure}../images/A2_2.png
:alt: A2_2.png
:width: 300px
```

En virtud de la simetría axial y de las condiciones de contorno equipotenciales a lo largo de los conductores interior ($r=R/2$) y exterior ($r=R$), la diferencia de potencial a través del espacio radial es idéntica en ambas regiones, inferior y superior. Como el campo eléctrico es estrictamente radial ($\vec{E}=E_r\hat{u}_r$) y la componente tangencial a lo largo de la interfaz vertical $z=h$ ($E_z=0$) es continua, el campo eléctrico radial $\vec{E}(r)$ es idéntico en ambos medios:

$$
\vec{E}(r)=-\frac{dV}{dr}\hat{u}_r
=\frac{V}{\ln\left(\frac{R}{R/2}\right)}\frac{1}{r}\hat{u}_r
=\frac{V}{\ln(2)r}\hat{u}_r
$$

Aunque $\vec{E}(r)$ permanece uniforme a través de la interfaz dieléctrica, el vector desplazamiento eléctrico $\vec{D}=\varepsilon(\vec{r})\vec{E}$ escala directamente con la permitividad local del material:

* **Región de dieléctrico líquido ($0\le z\le h$):**

  $$
  \vec{D}_{\varepsilon}(r)=\varepsilon\vec{E}(r)
  =\frac{\varepsilon V}{\ln(2)r}\hat{u}_r
  $$

* **Región de vacío ($h<z\le L$):**

  $$
  \vec{D}_{\varepsilon_0}(r)=\varepsilon_0\vec{E}(r)
  =\frac{\varepsilon_0V}{\ln(2)r}\hat{u}_r
  $$

La carga libre total $Q_l$ acumulada en el conductor interior se evalúa integrando el campo de desplazamiento $\vec{D}$ sobre la superficie del conductor en $r=R/2$:

$$
\begin{aligned}
Q_l
&=\iint \vec{D}\cdot d\vec{S} \\
&=\int_0^h D_{\varepsilon}\left(\frac{R}{2}\right)2\pi\left(\frac{R}{2}\right)dz
+\int_h^L D_{\varepsilon_0}\left(\frac{R}{2}\right)2\pi\left(\frac{R}{2}\right)dz
\end{aligned}
$$

Sustituyendo las magnitudes del desplazamiento se obtiene:

$$
\begin{aligned}
Q_l
&=\frac{2\pi\varepsilon hV}{\ln(2)}
+\frac{2\pi\varepsilon_0(L-h)V}{\ln(2)} \\
&=\frac{2\pi V}{\ln(2)}\left[\varepsilon h+\varepsilon_0(L-h)\right]
\end{aligned}
$$


```{note}
Esta capacidad dependiente de la altura del líquido se puede usar para resolver un sistema en el que el líquido tiene una altura variable, mediante la busqueda del punto de equilibrio. El sistema se comporta como dos condensadores en paralelo conectados a una diferencia de potencial $V$. La capacidad dependiente de la altura $C(h)$ es:

$$
\begin{aligned}
C(h)
&=\frac{Q_l}{V}
=C_{\text{dieléctrico}}(h)+C_{\text{vacío}}(L-h) \\
&=\frac{2\pi\varepsilon h}{\ln(2)}
+\frac{2\pi\varepsilon_0(L-h)}{\ln(2)}
\end{aligned}
$$

La energía electrostática total $W(h)$ almacenada en la geometría de campo es:

$$
W(h)=\frac{1}{2}C(h)V^2
=\frac{\pi V^2}{\ln(2)}\left[\varepsilon h+\varepsilon_0(L-h)\right]
$$

Como la energía almacenada depende explícitamente de la altura $h$ del fluido, una fuerza electrostática actúa a lo largo de la interfaz vertical. Bajo un potencial aplicado constante $V$, la fuerza electrostática neta hacia arriba $F_z$ ejercida sobre la columna de líquido dieléctrico se deriva del principio generalizado de los trabajos virtuales:

$$
F_z=+\left(\frac{\partial W}{\partial h}\right)_V
=\frac{1}{2}V^2\frac{dC(h)}{dh}
$$

Derivando $C(h)$ con respecto a $h$:

$$
\frac{dC(h)}{dh}=\frac{2\pi(\varepsilon-\varepsilon_0)}{\ln(2)}
$$

Sustituyendo esta derivada en la ecuación de fuerza se obtiene:

$$
F_z=\frac{\pi V^2(\varepsilon-\varepsilon_0)}{\ln(2)}
$$

Esta fuerza positiva hacia arriba atrae el fluido dieléctrico hacia el espacio entre electrodos. Microscópicamente, los campos de borde no uniformes en la superficie del líquido ejercen fuerzas sobre los dipolos moleculares inducidos, atrayendo el dieléctrico hacia regiones de campo eléctrico intenso para reducir la energía total del sistema a potencial constante $V$.

Esta fuerza en dirección $+z$ moverá el líquido hacia arriba hasta que la fuerza gravitatoria en direccion $-z$ que sufre el líquido elevado se iguale en módulo y se alcance un equilibrio.
```
































## S7. Materiales dieléctricos: Campo electrostático

Cuando la polarización es un dato, las densidades volumétrica y superficial de carga de polarización $\rho_P$ y $\sigma_P$ también son un dato.

$$
\rho_P &= -\nabla \cdot \vec{P} \\
\sigma_P &= \vec{P} \cdot \vec{n}
$$

Una vez tengo estas densidades de carga de polarización, el potencial $\varphi$ y el campo eléctrico $\vec{E}$ se calculan como en el tema 1, mediante integrales de carga, o si el sistema tuviera simetría, mediante el método de Gauss.

```{note}
**Al no existir proporcionalidad entre la polarización $\vec{P}$ y el campo eléctrico $\vec{E}$, no podemos hablar de permitividad del material**.
```

Observemos que, como en todos los materiales polarizados (*lineales o no*), la suma de la carga total debe ser nula.

$$
\int_V \rho_P(\vec{r}) dV + \oint \sigma_P(\vec{r}) dS = 0
$$


### Actividades prácticas

Calcular las densidades de carga de polarización y comprobar que la carga total de polarización es nula:

* Un cubo $x\in[0,a]$, $y\in[0,a]$ y $z\in[0,a]$ con una polarización constante $\vec{P} = P \vec{u_z}$.

```{figure}../images/A2_3.png
:alt: A2_3.png
:width: 250px
```

* Un cubo $x\in[0,a]$, $y\in[0,a]$ y $z\in[0,a]$ con una polarización constante $\vec{P} = \frac{P}{\sqrt{2}} (\vec{u_z}-\vec{u_y})$.

```{figure}../images/A2_4.png
:alt: A2_4.png
:width: 250px
```

* Un cubo $x\in[0,a]$, $y\in[0,a]$ y $z\in[0,a]$ con una polarización creciente en el eje $z$, $\vec{P} = P \frac{z}{a} \vec{u_z}$.

```{figure}../images/A2_5.png
:alt: A2_5.png
:width: 250px
```
